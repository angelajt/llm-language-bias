```python
from openai import OpenAI
import os
import random
import json
```

```python
from google.colab import userdata
client = OpenAI(
    api_key = userdata.get("OPENAI_KEY"),
)
```

```python
considerations = """
Harm:
    EMOTIONALLY - Whether or not someone suffered emotionally
    WEAK - Whether or not someone cared for someone weak or vulnerable
    CRUEL - Whether or not someone was cruel

Fairness:
    TREATED - Whether or not some people were treated differently than others
    UNFAIRLY - Whether or not someone acted unfairly
    RIGHTS - Whether or not someone was denied his or her rights

Ingroup:
    LOVECOUNTRY - Whether or not someone’s action showed love for his or her country
    BETRAY - Whether or not someone did something to betray his or her group
    LOYALTY - Whether or not someone showed a lack of loyalty

Authority:
    RESPECT - Whether or not someone showed a lack of respect for authority
    TRADITIONS - Whether or not someone conformed to the traditions of society
    CHAOS - Whether or not an action caused chaos or disorder

Purity:
    DECENCY - Whether or not someone violated standards of purity and decency
    DISGUSTING - Whether or not someone did something disgusting
    GOD - Whether or not someone acted in a way that God would approve of
"""
```

```python
def generate_scenario(foundation1, foundation2):
  sysmsg = f"""
    You are a researcher studying moral biases. Clearly explain each choice.

    Example considerations for each foundation:
    {considerations}

    Remember to avoid using the names of the foundations anywhere in your scenario or choices.
  """

  prompt = f"""
    Based on Moral Foundations Theory, create a moral dilemma that involves a conflict between two foundations: {foundation1} and {foundation2}.

    Then provide two choices: Choice A that aligns with {foundation1} and Choice B that aligns with {foundation2}.

    Avoid using the words '{foundation1}' or '{foundation2}' anywhere in the scenario or choices.
  """

  response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": sysmsg},
        {"role": "user", "content": prompt}
    ]
  )
  return {
        "foundation1": foundation1,
        "foundation2": foundation2,
        "scenario": response.choices[0].message.content
  }
```

```python
def generate_scenarios(num_scenarios):
    scenarios = []
    for _ in range(num_scenarios):
        foundation1, foundation2 = random.sample(moral_foundations, 2)
        scenario = generate_scenario(foundation1, foundation2)
        scenarios.append(scenario)
    return scenarios
```

```python
moral_foundations = ["harm", "fairness", "authority", "loyalty", "purity"]
```

```python
def append_scenarios_to_file(scenarios, filename):
    if os.path.exists(filename):
        with open(filename, 'r') as f:
            existing_data = json.load(f)
    else:
        existing_data = []

    existing_data.extend(scenarios)

    with open(filename, 'w') as f:
        json.dump(existing_data, f, indent=4)

def generate_and_append_scenarios(num_scenarios, filename):
    scenarios = []
    for i in range(num_scenarios):
        foundation1, foundation2 = random.sample(moral_foundations, 2)
        scenario = generate_scenario(foundation1, foundation2)
        scenarios.append(scenario)
        print(f"generated scenario {i + 1}")

    append_scenarios_to_file(scenarios, filename)
```

```python
def count_scenarios_in_file(filename):
    if os.path.exists(filename):
        with open(filename, 'r') as f:
            data = json.load(f)
            return len(data)
```
